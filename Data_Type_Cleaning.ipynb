```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/content/Employee.csv")
```

```python
display(df.head())
```

```python
df.info()
```

```python
print(df.dtypes)
```

```python
for column in df.columns:
    print(f"\n{column}:")
    print(df[column].unique())
```

```python
categorical_columns = [
    'Education',
    'City',
    'Gender',
    'EverBenched'
]

for column in categorical_columns:
    df[column] = df[column].astype('category')

numerical_columns = [
    'JoiningYear',
    'PaymentTier',
    'Age',
    'ExperienceInCurrentDomain',
    'LeaveOrNot'
]

for column in numerical_columns:
    df[column] = pd.to_numeric(df[column], errors='raise')

print("Data types converted successfully!")
```

```python
print("Cleaned Dataset:")
display(df.head(10))
```

```python
print("Corrected Data Types:")
print(df.dtypes)
```

```python
original_df = pd.read_csv('Employee.csv')

comparison = pd.DataFrame({
    'Original Data Type': original_df.dtypes.astype(str),
    'Corrected Data Type': df.dtypes.astype(str)
})

display(comparison)
```

```python
print("Missing Values:")
print(df.isnull().sum())
```
